```python
# Task 2.1

def total_value(price, quantity):
    """Calculate the total value of a product by multiplying its price by its quantity."""
    return price * quantity

print(total_value(1500, 40))
```

```python
# Task 2.2

def apply_discount(price, percent=10):
    """Return the price after applying a percentage discount."""
    return price * (1 - percent / 100)

print(apply_discount(1500))
print(apply_discount(1500, 25))
```

```python
# Task 2.3

def stock_status(quantity):
    """Return the stock status label based on quantity."""
    if quantity == 0:
        return "Out of stock"
    elif quantity < 10:
        return "Low stock"
    else:
        return "In stock"

print(stock_status(0))
print(stock_status(5))
print(stock_status(40))
```

```python
# Task 2.4

def products_in_category(inventory, category):
    """Return a list of product names belonging to the given category."""
    return [item["product"] for item in inventory if item["category"] == category]

print(products_in_category(inventory, "Electronics"))
```

**Task 4.5**

* **Rows and columns:** The DataFrame (`df`) has **11 rows** and **4 columns** (`(11, 4)` from `df.shape`).
   
* **Missing value:** The **`quantity`** column has **1** missing value (`NaN`), which belongs to the product **Highlighter Set** (row index `5`).

  
* **Duplicate row:** **Yes**, there is **1** duplicate row (`df.duplicated().sum()` returns `1`), which is **Pen** (row index `9` duplicates row index `1`: `Pen`, `Stationery`, `200`, `120.0`).

  
* **Data types (`price`, `category`, `quantity`):**
  * `price`: `int64`
  * `category`: `object` (strings in pandas)
  * `quantity`: `float64`
    
* **Why `quantity` is `float64`:** Because `quantity` contains a missing value (`None` / `NaN`), and standard `NaN` in pandas is a floating-point value. Pandas upcasts the entire column to `float64` so that `NaN` can be stored alongside the numeric quantities.

  
* **Average and highest price:** From `df.describe()`, the average (`mean`) price is **4427.27**, and the highest (`max`) price is **12000** (Backpack).

```python
def inspect_data(df):
    """Print the shape, missing value counts per column, and duplicate row count of a DataFrame."""
    print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")
    print("\nMissing values per column:")
    print(df.isnull().sum())
    print(f"\nDuplicate rows: {df.duplicated().sum()}")

inspect_data(df)
```
